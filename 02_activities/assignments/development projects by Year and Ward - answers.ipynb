
For each visualization, describe and justify:

What software did you use to create your data visualization?

  The visualizations for development projects by Year and Ward were created using Tableau Public. Tableau was selected because it enables rapid exploratory analysis, interactive dashboard creation, and clear aggregation of categorical and temporal data. Its drag-and-drop interface made it well suited for producing the year-based bar chart and ward-based trend line efficiently.

Who is your intended audience?
  The intended audience includes:

    - City of Toronto planners and policy analysts

    - Municipal housing and development decision-makers

    - Urban planning researchers and students

What information or message are you trying to convey with your visualization?
  The bar chart (Projects Received by Year) shows the growth trend in development applications over time, highlighting a significant increase in recent years followed by some fluctuation.

  The line chart (Projects Received by Ward) shows how development activity is unevenly distributed geographically, with certain wards experiencing substantially higher project volumes.

What aspects of design did you consider when making your visualization? How did you apply them? With what elements of your plots?

 Appropriate chart selection

 - A bar chart was used for yearly totals because it is effective for discrete time comparisons.

 - A line chart was used for ward comparisons to show variation across an ordered dimension.

  Clarity and readability

  I ensured:

   - Clear, descriptive titles

  - Labeled axes (Project Count, Year, Ward)

  - Consistent scaling

Color and visual hierarchy

  Used a neutral blue palette for the yearly bar chart to maintain a professional analytical tone.

  Used a contrasting purple line for ward comparison to visually separate the two analyses.

  Avoided overly bright or decorative colors.

  This improves focus and reduces cognitive load.

How did you ensure that your data visualizations are reproducible? If the tool you used to make your data visualization is not reproducible, how will this impact your data visualization?

 Reproducibility was supported through several steps:

  Used Tableau calculated fields (e.g., Year of Date Received) rather than manual preprocessing.

  Maintained consistent aggregation (COUNT of projects).

  Documented the data source and transformation logic.

  Ensured the dashboard can be regenerated by reconnecting to the same dataset.

However, Tableau workflows are less inherently reproducible than code-based tools because some steps are performed through the GUI.

How did you ensure that your data visualization is accessible?

 Color accessibility

  - Used distinct, non–red/green dependent colors.

  - Ensured sufficient contrast between marks and background.

Text readability

- Used legible font sizes.

- Avoided label overlap.

Cognitive accessibility

 - Limited the number of categories shown.

 - Used familiar chart types.

- Included clear titles and axis labels.
