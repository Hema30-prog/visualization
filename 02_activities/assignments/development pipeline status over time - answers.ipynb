What software did you use to create your data visualization?
  Python (pandas + matplotlib) to create the bar chart showing the number of projects by development pipeline status over time.
  Python was selected for its strong reproducibility and scripting capabilities.

Who is your intended audience?
  Municipal planners and housing policy analysts
  City of Toronto decision-makers

What information or message are you trying to convey with your visualization?
  Bar chart: shows trends in the volume of projects at different pipeline stages (Active, Built, Under Review) over time, helping identify growth patterns and pipeline pressure.

What aspects of design did you consider when making your visualization? How did you apply them? With what elements of your plots?

- Clarity and readability

  Used clear axis labels and descriptive titles.

  Rotated year labels for readability.

- Appropriate chart type

   Used a stacked bar chart to show part-to-whole relationships across pipeline stages.

- Color encoding

  Used consistent categorical colors for pipeline status.

How did you ensure that your data visualizations are reproducible? If the tool you used to make your data visualization is not reproducible, how will this impact your data visualization?

  Reproducibility was primarily ensured through Python scripting.

  Steps taken:

  Used a fully scripted pandas workflow.

  Converted dates programmatically.

  Created derived fields (Year, YearMonth) via code.

  Aggregated using explicit groupby operations.

  Avoided manual Excel manipulation.

  This means another analyst can reproduce the chart by running the same script on the same dataset.

How did you ensure that your data visualization is accessible?
- Color accessibility

  Used distinguishable categorical colors.

  Avoided red–green dependency.

  Used intensity gradients rather than hue-only encoding in the heatmap.

- Text readability

   Clear axis labels.
 
   Adequate font sizes.

   Rotated x-axis labels to prevent overlap.

- Non-color cues

  Included numeric axes and labels.

  Provided legend for categorical interpretation.

  Used descriptive titles.
