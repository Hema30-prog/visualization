```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

# Set random seed for reproducibility
np.random.seed(613)

# Make sample data
x = np.arange(50)
y = np.random.randint(0, 100,50)

# basic scatterplot
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y)

# Font dictionary
font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14} 

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1, loc ='left')
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()

fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()

```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y, marker='*', color ="indigo") # NEW ADDITIONS color = "indigo") # NEW ADDITIONS
fig.show()

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y, 
           marker='*', 
           color ="indigo", 
           linestyle = '--', 
           linewidth = 2) # NEW ADDITIONS color = "indigo") # NEW ADDITIONS
fig.show()

```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y,
        marker = '*',
        markersize = 12,
        color = '#7425b9',
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359',
        markerfacecolor = '#000000')
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y,
        marker = '*',
        markersize = 12,
        color = '#7425b9',
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359',
        markerfacecolor = '#000000')
ax.grid(axis = 'y', color ="green", linewidth = 2, linestyle = '--')
```
